```python
# 1️⃣ Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# For better chart display
plt.style.use('seaborn-v0_8')
sns.set_palette('viridis')

# 2️⃣ Load Dataset
df = pd.read_csv("advanced_employee_data.csv")

# Show basic info
print("✅ Data Loaded Successfully!\n")
print(df.head())
print("\nDataset Shape:", df.shape)

# 3️⃣ Basic Data Insights
print("\n--- Basic Info ---")
print(df.info())

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Statistical Summary ---")
print(df.describe())

# 4️⃣ Department-Wise Employee Count
plt.figure(figsize=(8, 5))
sns.countplot(x='Department', data=df)
plt.title("Employees per Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()

# 5️⃣ Average Salary by Department
plt.figure(figsize=(8, 5))
dept_salary = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)
sns.barplot(x=dept_salary.index, y=dept_salary.values)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (₹)")
plt.xticks(rotation=45)
plt.show()

# 6️⃣ Gender Distribution
plt.figure(figsize=(5, 5))
df['Gender'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, colors=['#66b3ff','#ff9999'])
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()

# 7️⃣ Age vs Salary Relationship
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Age', y='Salary', hue='Department', data=df)
plt.title("Age vs Salary by Department")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()

# 8️⃣ Correlation Heatmap (Only Numeric Columns)
plt.figure(figsize=(8,6))
corr = df.select_dtypes(include=['int64', 'float64']).corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Employee Metrics Correlation Heatmap")
plt.show()

# 9️⃣ Top 10 Highest Paid Employees
print("\n--- Top 10 Highest Paid Employees ---")
print(df[['Name', 'Department', 'Salary']].sort_values(by='Salary', ascending=False).head(10))

# 10️⃣ Experience vs Salary Trend
plt.figure(figsize=(8,5))
sns.lineplot(x='Experience_Years', y='Salary', data=df)
plt.title("Experience vs Salary Trend")
plt.xlabel("Years of Experience")
plt.ylabel("Salary (₹)")
plt.show()

print("\n🎯 Analysis Completed Successfully!")
```
